Sorting is one of the most fundamental operations in computer science.
This notebook covers all major sorting algorithms with visualizations.

**Time Complexity Summary:**
| Algorithm | Best | Average | Worst | Space |
|-----------|------|---------|-------|-------|
| Bubble | O(n) | O(n²) | O(n²) | O(1) |
| Merge | O(n logn) | O(n logn) | O(n logn) | O(n) |
| Quick | O(n logn) | O(n logn) | O(n²) | O(logn) |
| Heap | O(n logn) | O(n logn) | O(n logn) | O(1) |

```python
import matplotlib.pyplot as plt, time, random
algos = {'Bubble':bubble_sort,'Merge':merge_sort,'Quick':quick_sort,'Heap':heap_sort}
sizes = [50,100,200,500,1000]
plt.figure(figsize=(12,6))
for name, fn in algos.items():
    times = []
    for n in sizes:
        arr = random.sample(range(n*5), n)
        t0 = time.perf_counter(); fn(arr); times.append((time.perf_counter()-t0)*1000)
    plt.plot(sizes, times, marker='o', label=name)
plt.xlabel('n'); plt.ylabel('Time (ms)'); plt.title('Sorting Performance Comparison')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

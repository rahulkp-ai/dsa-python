# Heaps & Priority Queues

## Heap Data Structure

A complete binary tree where parent >= children (max-heap) or parent <= children (min-heap).

**Operations:** push O(logn), pop O(logn), peek O(1), build O(n)

**Applications:** Priority queues, scheduling, Dijkstra, top-K problems

# Bit Manipulation — Tricks & Patterns

## Bit Operations

| Op | Symbol | Example |
|----|--------|---------|
| AND | & | 5&3=1 |
| OR | \| | 5\|3=7 |
| XOR | ^ | 5^3=6 |
| NOT | ~ | ~5=-6 |
| Left Shift | << | 5<<1=10 |
| Right Shift | >> | 5>>1=2 |

**Key tricks:**
- `n & (n-1)` — clears lowest set bit
- `n & (-n)` — isolates lowest set bit
- `n ^ n = 0` — XOR with self
- `n ^ 0 = n` — XOR with zero

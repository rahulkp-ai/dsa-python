# Graph Algorithms — BFS, DFS, Shortest Paths

## Graph Theory Fundamentals

A graph G = (V, E) where V = vertices, E = edges.

**Types:**
- Directed / Undirected
- Weighted / Unweighted
- Cyclic / Acyclic (DAG)

**Representations:**
- Adjacency List: O(V+E) space — use for sparse graphs
- Adjacency Matrix: O(V²) space — use for dense graphs

```python
# BFS vs DFS
g = {0:[1,2],1:[3,4],2:[5],3:[],4:[],5:[]}
print('BFS:', bfs(g,0))
print('DFS:', dfs(g,0))
print('Shortest 0->5:', bfs_shortest_path(g,0,5))
```

```python
# Dijkstra
wg={0:[(1,4),(2,1)],1:[(3,1)],2:[(1,2),(3,5)],3:[]}
print('Dijkstra:', dijkstra(wg,0))
```

```python
# Topological Sort
print('Topo:', topological_sort(6,[[5,2],[5,0],[4,0],[4,1],[2,3],[3,1]]))
# Union-Find
uf=UnionFind(5); uf.union(0,1); uf.union(1,2)
print('Connected(0,2):', uf.connected(0,2))
print('Components:', uf.components)
```

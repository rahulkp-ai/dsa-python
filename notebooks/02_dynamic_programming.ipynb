# Dynamic Programming — Patterns & Problems

## Introduction to Dynamic Programming

DP = Recursion + Memoization = Optimal Substructure + Overlapping Subproblems

**When to use DP:**
- Optimization problem (max/min)
- Counting problem
- Decision problem (yes/no)
- Problem has optimal substructure

**Two approaches:**
1. **Top-down (Memoization)** — Recursive + cache
2. **Bottom-up (Tabulation)** — Iterative DP table

```python
# Knapsack
print(knapsack_01([2,3,4,5],[3,4,5,6],5))
# Trace: dp table
n=4; cap=5; w=[2,3,4,5]; v=[3,4,5,6]
dp=[[0]*(cap+1) for _ in range(n+1)]
for i in range(1,n+1):
    for wt in range(cap+1):
        dp[i][wt]=dp[i-1][wt]
        if w[i-1]<=wt: dp[i][wt]=max(dp[i][wt],dp[i-1][wt-w[i-1]]+v[i-1])
print('DP Table:'); [print(row) for row in dp]
```
